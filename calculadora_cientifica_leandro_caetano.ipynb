```python
import math

def mostrar_menu():
    print("\n===== CALCULADORA CIENTÍFICA =====")
    print("1  - Soma")
    print("2  - Subtração")
    print("3  - Multiplicação")
    print("4  - Divisão")
    print("5  - Raiz Quadrada")
    print("6  - Potência")
    print("7  - Logaritmo Natural (ln)")
    print("8  - Logaritmo Base 10")
    print("9  - Fatorial")
    print("10 - Arredondar para Cima (ceil)")
    print("11 - Arredondar para Baixo (floor)")
    print("12 - Seno (sin)")
    print("13 - Cosseno (cos)")
    print("14 - Tangente (tan)")
    print("15 - Converter para Graus")
    print("16 - Converter para Radianos")
    print("17 - Exponencial (exp)")
    print("18 - Valor Absoluto")
    print("19 - Separar Parte Inteira e Fracionária")
    print("20 - Resto da Divisão (mod)")
    print("21 - Máximo entre dois números")
    print("22 - Mínimo entre dois números")
    print("0  - Sair")


while True:
    mostrar_menu()

    try:
        opcao = int(input("\nEscolha uma operação: "))

        if opcao == 0:
            print("Encerrando a calculadora...")
            break

        elif opcao == 1:
            a = float(input("Primeiro número: "))
            b = float(input("Segundo número: "))
            print("Resultado:", a + b)

        elif opcao == 2:
            a = float(input("Primeiro número: "))
            b = float(input("Segundo número: "))
            print("Resultado:", a - b)

        elif opcao == 3:
            a = float(input("Primeiro número: "))
            b = float(input("Segundo número: "))
            print("Resultado:", a * b)

        elif opcao == 4:
            a = float(input("Primeiro número: "))
            b = float(input("Segundo número: "))
            if b == 0:
                print("Erro: divisão por zero!")
            else:
                print("Resultado:", a / b)

        elif opcao == 5:
            a = float(input("Digite um número: "))
            print("Resultado:", math.sqrt(a))

        elif opcao == 6:
            a = float(input("Base: "))
            b = float(input("Expoente: "))
            print("Resultado:", math.pow(a, b))

        elif opcao == 7:
            a = float(input("Número: "))
            print("Resultado:", math.log(a))

        elif opcao == 8:
            a = float(input("Número: "))
            print("Resultado:", math.log10(a))

        elif opcao == 9:
            a = int(input("Número inteiro: "))
            print("Resultado:", math.factorial(a))

        elif opcao == 10:
            a = float(input("Número: "))
            print("Resultado:", math.ceil(a))

        elif opcao == 11:
            a = float(input("Número: "))
            print("Resultado:", math.floor(a))

        elif opcao == 12:
            a = float(input("Ângulo em radianos: "))
            print("Resultado:", math.sin(a))

        elif opcao == 13:
            a = float(input("Ângulo em radianos: "))
            print("Resultado:", math.cos(a))

        elif opcao == 14:
            a = float(input("Ângulo em radianos: "))
            print("Resultado:", math.tan(a))

        elif opcao == 15:
            a = float(input("Radianos: "))
            print("Resultado:", math.degrees(a))

        elif opcao == 16:
            a = float(input("Graus: "))
            print("Resultado:", math.radians(a))

        elif opcao == 17:
            a = float(input("Número: "))
            print("Resultado:", math.exp(a))

        elif opcao == 18:
            a = float(input("Número: "))
            print("Resultado:", math.fabs(a))

        elif opcao == 19:
            a = float(input("Número: "))
            fracao, inteiro = math.modf(a)
            print("Parte inteira:", inteiro)
            print("Parte fracionária:", fracao)

        elif opcao == 20:
            a = float(input("Número 1: "))
            b = float(input("Número 2: "))
            print("Resto:", a % b)

        elif opcao == 21:
            a = float(input("Número 1: "))
            b = float(input("Número 2: "))
            print("Maior número:", max(a, b))

        elif opcao == 22:
            a = float(input("Número 1: "))
            b = float(input("Número 2: "))
            print("Menor número:", min(a, b))

        else:
            print("Opção inválida!")

    except ValueError:
        print("Erro: digite um número válido!")
```
